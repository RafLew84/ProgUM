# Lista 1 - Nawigacja w Jetpack Compose

**Termin wykonania:** 2 tygodnie od pierwszych zajęć. Dokładne daty dla poszczególnych grup zostaną podane na zajęciach.

**Zadanie praktyczne: 6 pkt. Odpowiedź ustna: 4 pkt. Łącznie: 10 pkt.**

## Zadanie 1 - ComposeShop (6 pkt)

Zaimplementuj aplikację ComposeShop w Kotlinie i Jetpack Compose. Zastosuj Navigation Compose 2, typowane trasy z adnotacją `@Serializable`, jedną Activity, jeden `NavController` i jeden główny `NavHost`. Proces zamówienia umieść w grafie zagnieżdżonym Checkout.

Dane produktów, profilu i historii zamówień przygotuj lokalnie. Koszyk może zawierać stały zestaw produktów. Płatność jest symulowana. Nie są wymagane API, baza danych, logowanie ani integracja płatności. Wygląd interfejsu jest dowolny, ale treść i nawigacja mają być czytelne.

### 1. Ekrany główne i dane - 1 pkt

- Przygotuj co najmniej 6 produktów z unikatowymi identyfikatorami, nazwami, cenami i kategoriami. Uwzględnij co najmniej 2 kategorie.
- Dolny pasek ma 4 pozycje: Start (lista produktów), Kategorie (wybór kategorii i filtrowanie produktów), Koszyk oraz Ulubione.
- Umożliwiaj oznaczanie produktów jako ulubione i usuwanie tego oznaczenia. Zmiany mają być widoczne na ekranie Ulubionych także po przełączeniu zakładki.
- Koszyk pokazuje produkty i ich łączną cenę oraz przycisk „Przejdź do zamówienia”.

### 2. Typowane trasy i szczegóły produktu - 1 pkt

- Kliknięcie produktu otwiera ekran szczegółów. Trasa tego ekranu ma argument id typu Int.
- Przekaż do nawigacji identyfikator produktu i odczytaj go przez `toRoute<Details>()`. Na podstawie ID wybierz dane z lokalnej kolekcji.
- Ekran szczegółów pokazuje co najmniej nazwę, cenę i kategorię. Powrót prowadzi do ekranu, z którego otwarto szczegóły.
- Wszystkie cele i graf Checkout definiuj jako typowane trasy. Do ekranu szczegółów przekazuj ID, a nie cały obiekt produktu.

### 3. Szuflada i powiązanie menu z nawigacją - 1 pkt

- Szuflada zawiera pozycje: Profil (przykładowe dane użytkownika), Historia zamówień (przykładowa lista z datami i statusami) oraz Ustawienia (co najmniej jeden działający przełącznik).
- Otwieraj szufladę przyciskiem menu. Po wybraniu pozycji przejdź do odpowiedniego ekranu i zamknij szufladę.
- Szuflada i dolny pasek korzystają ze wspólnego `NavController`. Aktywną pozycję menu wyznaczaj na podstawie aktualnego celu nawigacji, a nie osobnej zmiennej `selected`.

### 4. Graf zagnieżdżony Checkout - 1 pkt

- Przycisk w koszyku otwiera graf Checkout. Jego cel początkowy to Address.
- Address: formularz z odbiorcą, ulicą i numerem oraz kodem pocztowym i miastem. Przycisk „Dalej” prowadzi do Payment.
- Payment: wybór jednej z co najmniej 3 metod płatności. Przycisk „Podsumowanie” prowadzi do Summary.
- Summary: produkty z koszyka, łączna cena, adres dostawy i wybrana metoda płatności. Przycisk „Potwierdź zamówienie” kończy proces.

### 5. Stan formularza i walidacja - 1 pkt

- Nie pozwalaj przejść dalej z pustymi wymaganymi polami adresu ani bez wybranej metody płatności. Wystarczy sprawdzanie, czy pola nie są puste po `trim()`; pokaż czytelną informację o błędzie.
- Wpisane dane i wybór płatności mają pozostać po cofnięciu do poprzedniego kroku, ponownym przejściu dalej oraz obrocie urządzenia.
- Przechowuj wspólny stan ponad ekranami, które z niego korzystają. Funkcje ekranów otrzymują wartości i callbacki aktualizujące dane lub zgłaszające akcje.
- Po potwierdzeniu wyczyść dane formularza. Ponowne rozpoczęcie zamówienia nie może przywrócić poprzedniego adresu.

### 6. Stos nawigacji - 1 pkt

- Wielokrotne wybranie tej samej zakładki nie dodaje kolejnych kopii ekranu.
- W Checkout ukryj dolny pasek i ikonę otwierania szuflady, wyłącz gest jej otwierania oraz zamknij szufladę, jeśli była otwarta.
- Wstecz prowadzi z Summary do Payment, z Payment do Address, a z Address do koszyka. Przycisk w aplikacji i systemowy Wstecz mają zachowywać się spójnie.
- Potwierdzenie pokazuje informację o przyjęciu zamówienia, usuwa graf Checkout wraz z jego ekranami ze stosu i otwiera Start. Kolejne użycie Wstecz nie może otworzyć adresu, płatności ani podsumowania zakończonego zamówienia. Wcześniejsza historia głównej aplikacji może pozostać.

### Punktacja zadania i prezentacja rozwiązania

Każda z sześciu części zadania jest oceniana osobno, maksymalnie na 1 pkt.

Na laboratorium uruchom projekt na emulatorze lub urządzeniu i pokaż: szczegóły dwóch różnych produktów, zmianę zakładek i pozycji szuflady, przejście przez zamówienie z cofnięciem, zachowanie danych po obrocie oraz działanie Wstecz po potwierdzeniu zamówienia. Przygotuj kod do omówienia.
