# Lista 2 - Korutyny i współbieżne pobieranie danych

## Zadanie 1 - Kino GURU (4 pkt)

Zaimplementuj aplikację Kino GURU w Kotlinie i Jetpack Compose. Aplikacja ma dwa ekrany: listę popularnych filmów i szczegóły wybranego filmu. Dane przygotuj lokalnie, a oczekiwanie na nie symuluj przez `delay()`. Wygląd jest dowolny, ale stan aplikacji i wyniki muszą być czytelne.

### 1. Lokalne dane i funkcje suspend - 1 pkt

- Przygotuj co najmniej 10 filmów o unikatowych identyfikatorach. Dla każdego filmu zapewnij tytuł, reżysera, rok, gatunek i krótki opis, co najmniej 2 recenzje oraz co najmniej 3 podobne filmy z tej samej kolekcji. Wybrany film nie może polecać samego siebie.
- Zaimplementuj cztery funkcje `suspend` zgodne z kontraktami poniżej. Dane szczegółowe muszą zależeć od przekazanego `movieId`. Modele `Movie`, `MovieDetails` i `Review` zdefiniuj samodzielnie.
- `fetchPopularMovies(): List<Movie>` - po `delay(2000)` zwraca listę filmów.
- `fetchMovieDetails(movieId: Int): MovieDetails` - po `delay(1500)` zwraca dane podstawowe i opis.
- `fetchMovieReviews(movieId: Int): List<Review>` - w udanej próbie po `delay(3000)` zwraca recenzje.
- `fetchSimilarMovies(movieId: Int): List<Movie>` - po `delay(2000)` zwraca podobne filmy.

### 2. Lista filmów i nawigacja - 1 pkt

- Po wejściu na ekran główny rozpocznij pobieranie filmów w korutynie. W czasie oczekiwania pokaż `CircularProgressIndicator`, następnie listę w `LazyColumn`.
- Kliknięcie filmu przechodzi na ekran szczegółu. Przekaż identyfikator `Int` w typowanej trasie i odczytaj go przez `toRoute`. Przycisk Wstecz oraz systemowy Wstecz mają wracać do listy.
- Pobieranie nie może uruchamiać się od nowa przy każdej rekompozycji. Uruchomienie przy wejściu na ekran możesz powiązać z `LaunchedEffect`, a obsługę przycisków z `rememberCoroutineScope()` i `launch`.
- Interfejs musi reagować podczas oczekiwania. Do symulacji opóźnień używaj `delay()`.

### 3. Współbieżne ładowanie szczegółów - 1 pkt

- Wejście na ekran szczegółów automatycznie rozpoczyna pobieranie w trybie współbieżnym. Jedna operacja ładowania obejmuje dane podstawowe, recenzje i podobne filmy.
- Przygotuj funkcję `suspend` łączącą trzy wyniki. Wewnątrz `coroutineScope { ... }` uruchom trzy korutyny `async`: najpierw utwórz wszystkie zadania, a potem odbierz ich wyniki przez `await()`.
- Wszystkie trzy zadania muszą należeć do tej samej hierarchii rodzic-dzieci. Nie twórz dla nich niezależnych zakresów i nie używaj `GlobalScope`.
- Po sukcesie pokaż komplet danych w sekcjach „Opis”, „Recenzje” i „Podobne filmy”. Nie pokazuj częściowego zestawu jako ukończonego wyniku.

### 4. Porównanie trybów i pomiar czasu - 1 pkt

- Dodaj wybór trybu „Współbieżnie” lub „Sekwencyjnie” oraz przycisk „Wczytaj ponownie”. Zmiana wyboru dotyczy kolejnej próby. W trybie sekwencyjnym wywołaj te same trzy funkcje `suspend` kolejno, bez `async`.
- Po każdej udanej próbie pokaż zastosowany tryb i rzeczywiście zmierzony czas uzyskania całego zestawu szczegółów. Mierz od rozpoczęcia pobierania szczegółów do uzyskania wszystkich wyników; nie wliczaj wcześniejszego pobierania listy filmów.
- Dla udanej próby oczekujemy około 3 s współbieżnie i około 6,5 s sekwencyjnie. Są to wartości orientacyjne, zależne od urządzenia i planowania zadań; nie wpisuj ich na stałe do UI.
- Użyj zegara monotonicznego, np. `kotlin.time.measureTimedValue` lub `TimeSource.Monotonic`. W obu trybach aplikacja ma pozostać responsywna.
